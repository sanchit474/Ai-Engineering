```python
# Stemming is a text preprocessing technique in NLP where words are reduced to their root/base form by removing prefixes or suffixes.
```

```python
#  classification problem
# comment of a blog post is positive or negative
# we can use stemming to reduce the number of unique words in our dataset, which can help
words = ["run", "running", "runner", "ran", "runs", "write", "writing", "writer", "wrote", "writes", "program", "programming", "programmer", "programmed", "programs","history"]
words
```

# NLTK provides the PorterStemmer.(basic but not accurate ans opt)

```python
from nltk.stem import PorterStemmer
```

```python
stemming = PorterStemmer()
for word in words:
    print( word + " -> " + stemming.stem(word))  
```

```python
# Stemming may produce words that are not actual English words: History -> histori
# because it simply cuts suffixes using rules.

# Common Stemmers in NLTK
# PorterStemmer() → most commonly used
# LancasterStemmer() → more aggressive
# SnowballStemmer() → improved version of PorterStemmer, supports multiple languages
```

# RegexpStemmer  (opt)

```python
# RegexpStemmer() → allows you to define your own stemming rules using regular expressions
from nltk.stem import RegexpStemmer
regexp_stemmer = RegexpStemmer('ing$|s$||ed$|er$|ly$')
for word in words:
    print( word + " -> " + regexp_stemmer.stem(word))   
# if in the end end $ is used then it will only remove the suffix if it is at the end of the word. otherwise it will remove it from anywhere in the word.
```

SnowballStemmer (support more languages and better op then porterstemmer| opt)

```python
# SnowballStemmer() → improved version of PorterStemmer, supports multiple languages
from nltk.stem import SnowballStemmer
snowball_stemmer = SnowballStemmer('english')
for word in words:
    print( word + " -> " + snowball_stemmer.stem(word))
```
