```python
paragraph = """ I have three visions for India. In 3000 years of our history, people from all over the world have come and invaded us, captured our lands, conquered our minds. From Alexander onwards, the Greeks, the Turks, the Moguls, the Portuguese, the British, the French, the Dutch, they came and looted us, took over what was ours. Yet we have not done this to any other nation. We have not conquered anyone. We have not grabbed their land, their culture, their history and tried to enforce our way of life on them. Why? Because we respect the freedom of others.That is why my first vision is that of freedom. I believe that India got its first vision of this in 1857, when we started the war of independence. It is this freedom that we must protect and nurture and build on. If we are not free, no one will respect us. My second vision for India's development. For fifty years we have been a developing nation. It is time we see ourselves as a developed nation. We are among the top 5 nations of the world in terms of GDP. We have a 10 percent growth rate in most areas. Our poverty levels are falling. Our achievements are being globally recognised today. Yet we lack the self-confidence to call ourselves a developed nation, or to claim our rightful place in the world arena. I have a third vision. India must stand up to the world. Because I believe that unless India stands up to the world, no one will respect us. Only strength respects strength. We must be strong not only as a military power but also as an economic power. Both must go hand-in-hand. My good fortune was to have worked with three great minds. Dr Vikram Sarabhai of the Dept of space, Professor Satish Dhawan who succeeded him and Dr Abdul Kalam who succeeded him as the head of ISRO. I was lucky to have worked with all three of them closely and consider myself fortunate to have been mentored by such great minds."""
# paragraph
```

```python
import nltk
nltk.download('stopwords') # for stop words
nltk.download('punkt')  # for tokenization
nltk.download('wordnet')  # for lemmatization
```

```python
from nltk.corpus import stopwords
stopwords.words('english')
```

```python
# how we gonna to use stop words
#  the flow would be like this
# 1. tokenize the paragraph into sentences/words
# 2. remove the stop words from the list of words
```

stop_words + porter_stemmer + word_tokenize

```python
# go with stemming first
# apply stopwords and filter then apply stemming and lemmatization
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
```

```python

from nltk.tokenize import sent_tokenize
sentences = sent_tokenize(paragraph)
```

```python
from nltk.tokenize import word_tokenize
```

```python
for i in range(len(sentences)):
    words = word_tokenize(sentences[i])
    words = [stemmer.stem(word) for word in words if word not in stopwords.words('english')] #removing stop words and applying stemming
    sentences[i] = ' '.join(words) #converting list of words to sentence
sentences
```

# using snowball stemmer + stop words

```python
from nltk.stem import SnowballStemmer
snowball_stemmer = SnowballStemmer('english')
```

```python
for i in range(len(sentences)):
    words = word_tokenize(sentences[i])
    words = [snowball_stemmer.stem(word) for word in words if word not in stopwords.words('english')] #removing stop words and applying stemming
    sentences[i] = ' '.join(words) #converting list of words to sentence
sentences
```

# stopwords and filter then apply lemmatization

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
```

```python
for i in range(len(sentences)):
    words = word_tokenize(sentences[i])
    words = [lemmatizer.lemmatize(word) for word in words if word not in stopwords.words('english', 'v')] #removing stop words and applying lemmatization
    sentences[i] = ' '.join(words) #converting list of words to sentence
sentences
```
