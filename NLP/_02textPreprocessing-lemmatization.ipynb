# Lemmatization 
Lemmatization is a NLP technique that converts a word into its meaningful dictionary root form, called a lemma.
Unlike stemming, it uses:

vocabulary
grammar
part-of-speech (POS)

So the output is usually a valid English word.

```python
words = ["run", "running", "runner", "ran", "runs", "write", "writing", "writer", "wrote", "writes", "program", "programming", "programmer", "programmed", "programs","history"]
words
```

```python
import nltk
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
for word in words:
    print( word + " -> " + lemmatizer.lemmatize(word, pos='v') )    
```
