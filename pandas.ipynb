```python
df.where(df['Age'] >= 30) # filter rows where Age is greater than 30 and replace other values with NaN
```

```python
df.where(df['Age'] > 30, other='Not Eligible') # filter rows where Age is greater than 30 and replace other values with 'Not Eligible'
```

```python
# 5 Rows and Columns - Operation (Add, update, delete)

# 5.1.1 Adding a new column to the data frame
# df
df['Country'] = ['USA', 'USA', 'USA'] # add a new column to the data frame
df
```

```python
# 5.1.2 Adding a new column to the data frame using existing column/s

df['serving_period']= 60-df['Age'] # add a new column to the data frame using existing column
df
```

```python
#5.1.3 adding new row to the data frame

df.loc[len(df)] = ['David', 40, 'Houston', 'USA', 20] # add a new row to the data frame using loc
df
```

```python
# 5.2.1 updating values in the data frame

df.loc[df['Name'] == 'Alice', 'Age'] = 35  # update the value in the 'Age' column for Alice using column name
df.loc[1, ['City', 'serving_period']] = ['San Francisco', 30] # update multiple values in the data frame using loc index and column name
df
```

```python
#5.3.1 delete value - rows and columns

df = df.drop(df[df['Name'] == 'David'].index) # delete row where Name is David
df
```

```python
# 5.3.2  delete row using index-name filter

df.drop(2, axis=0, inplace=True) # delete row with index 2 using drop method and inplace parameter to modify the original data frame
df
```

```python
# 5.3.3 delete column using column name

df.drop('Country', axis=1, inplace=True) # delete column 'Country' using drop method and inplace parameter to modify the original data frame
df
```

```python
# 5.4.1rename column name

df.rename(columns={'serving_period': 'Years_to_Serve'}, inplace=True) # rename column name using rename method and inplace parameter to modify the original data frame
df
```

```python
# 6.1 sorting data descending order by Age column

df.sort_values(by='Age', ascending=False, inplace=True) # sort data frame by Age column in descending order using sort_values method and inplace parameter to modify the original data frame
df
```

```python
# 6.2 sorting data ascending order by Age column

df.sort_values(by='Age', ascending=True, inplace=True) # sort data frame by Age column in ascending order using sort_values method and inplace parameter to modify the original data frame
df
```
