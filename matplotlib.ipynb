```python
# import matplotlib as mpl
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [6,4,6,2,10]

plt.plot(x, y) # Plotting the points
plt.title("Simple Plot") # Title for the plot
plt.xlabel("X-axis") # Label for x-axis
plt.ylabel("Y-axis") # Label for y-axis

plt.show() # Display the plot
```

```python
# style , color , marker , linestyle accordingly to your need
plt.plot(x, y, color='red', marker='o', linestyle='--') # Customizing the plot
plt.title("Customized Plot") # Title for the plot
plt.show
```

```python
# Advance line charts- multiple lines and legends
x = [1,2,3,4,5]
y1 = [6,4,6,2,10]
y2 = [2,3,4,5,6]
plt.plot(x, y1, color='red', marker='o', linestyle='--', label='Line 1') # First line
plt.plot(x, y2, color='blue', marker='s', linestyle='-', label='Line 2') # Second line
# plt.title("Multiple Lines Plot") # Title for the plot
# plt.xlabel("X-axis") # Label for x-axis
# plt.ylabel("Y-axis") # Label for y-axis
# plt.legend() # Show legend
plt.show() # Display the plot
```

```python
# Bar charts
categories = ['A', 'B', 'C', 'D', 'E']
values = [10, 15, 7, 12, 20]
plt.bar(categories, values, color='green') # Creating a bar chart
plt.title("Bar Chart") # Title for the plot

plt.xlabel("Categories") # Label for x-axis
plt.ylabel("Values") # Label for y-axis
plt.show() # Display the plot
```
