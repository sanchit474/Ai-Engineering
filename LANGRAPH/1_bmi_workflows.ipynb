```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
# define state                       type for BMI workflow
class BMIState(TypedDict):
    height_m: float
    weight_kg: float
    bmi: float
    category: str
```

```python
def calculate_bmi(state: BMIState) -> BMIState:
    """Calculate BMI given height in meters and weight in kilograms."""
    weight = state["weight_kg"]
    height = state["height_m"]

    bmi = weight / (height ** 2)
    state["bmi"] = round(bmi, 2)
    return state
```

```python
def label_bmi(state: BMIState) -> BMIState:
    bmi = state["bmi"]
    if bmi < 18.5:
        state["category"] = "Underweight"
    elif 18.5 <= bmi < 24.9:
        state["category"] = "Normal weight"
    elif 25 <= bmi < 29.9:
        state["category"] = "Overweight"
    else:
        state["category"] = "Obesity"
    return state
```

```python
# define the BMI workflow graph
graph = StateGraph(BMIState)

# add nodes to the graph
graph.add_node("calculate_bmi", calculate_bmi) 
graph.add_node('label_bmi', label_bmi)
# define the workflow edges
graph.add_edge(START, "calculate_bmi")
graph.add_edge("calculate_bmi", 'label_bmi')
graph.add_edge('label_bmi', END)

#compile the graph to create a workflow
workflow = graph.compile()
```

```python
initial_state = BMIState(height_m=1.75, weight_kg=70, bmi=0.0)
# execute the workflow
final_state = workflow.invoke(initial_state)
print(final_state)  # Output: {'height_m': 1.75, 'weight_kg': 70, 'bmi': 22.86}
```
